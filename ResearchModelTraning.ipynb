```python
import pandas as pd
import json

# Load JSON properly
with open("C:/Users/vivek/Downloads/acndata_sessions.json") as f:
    data = json.load(f)

# Convert to dataframe
df = pd.json_normalize(data["_items"])

df.head()
```

```python
df = df[['connectionTime', 'kWhDelivered']]
df.dropna(inplace=True)

df.head()
```

```python
df['connectionTime'] = pd.to_datetime(df['connectionTime'])
df['hour'] = df['connectionTime'].dt.hour

df.head()
```

```python
data = df.groupby('hour').size().reset_index(name='vehicles')

data.head()
```

```python
from sklearn.linear_model import LinearRegression

X = data[['hour']]
y = data['vehicles']

model = LinearRegression()
model.fit(X, y)

print("Model trained successfully!")
```

```python
import pickle

pickle.dump(model, open("model.pkl", "wb"))

print("Model saved!")
```
